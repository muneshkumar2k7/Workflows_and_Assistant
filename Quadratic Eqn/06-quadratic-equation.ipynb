```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict, Literal
```

```python
class QuadState(TypedDict):
    a: int 
    b: int 
    c: int

    equation : str
    discriminant : float
    result : str
```

```python
graph=StateGraph(QuadState)

def show_equation(state : QuadState)->QuadState:
     state['equation'] = f'{state["a"]} x^2 +  {state["b"]}x  + {state["c"]}'     
     return state


def calc_discriminant(state:QuadState)->QuadState:
    a = state['a']
    b = state['b']
    c = state['c']

    state['discriminant'] = b*b - 4*a*c

    return state


def real_roots(state : QuadState)->QuadState:

    root1 =  (- state['b']  - (state['discriminant'])**0.5) / (2 * state['a'])
    root2 =  (- state['b']  + (state['discriminant'])**0.5) / (2 * state['a'])
    
    result = f"real_roots are {root1} and {root2}"

    state['result'] = result

    return state 
    
      

def one_real_root(state : QuadState)->QuadState:
    root = - state['b'] / (2 * state['a'])
    
    result =  f"one_real_root is {root}" 
    state['result'] = result 

    return  state


def no_real_root(state : QuadState)->QuadState:
    result = "no_real_roots" 
    state['result'] =  result

    return state
    





def check_condition(state : QuadState)->Literal["real_roots","no_real_root","one_real_root"]:
      d = state["discriminant"]

      if d>0:
           return "real_roots"
      elif d==0:
           return "one_real_root"
      else:
           return "no_real_root"
```

```python
graph = StateGraph(QuadState)

graph.add_node('show_equation',show_equation)
graph.add_node('real_roots',real_roots)
graph.add_node('no_real_roots',no_real_root)
graph.add_node('one_real_root',one_real_root)
graph.add_node('calculate_discriminant',calc_discriminant)


graph.add_edge(START, "show_equation")
graph.add_edge("show_equation", "calculate_discriminant")
graph.add_conditional_edges("calculate_discriminant",check_condition)
graph.add_edge("one_real_root",END)
graph.add_edge("real_roots",END)
graph.add_edge("no_real_root",END)

workflow =graph.compile()

```
